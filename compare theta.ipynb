```python
df1 = pd.read_pickle("DMSO/traj_224K_IRspectrum_after_equilibration/theta histogram.pkl")
df2 = pd.read_pickle("DMSO_9/traj_224K_IR_after_eq/theta histogram.pkl")
df3 = pd.read_pickle("DMSO_61/traj_224K_IR_after_eq/theta histogram.pkl")
df4 = pd.read_pickle("DMSO/traj_224K_IRspectrum/theta histogram.pkl")
```

```python
theta1 = df1.loc[0, "theta"]; rho1 = df1.loc[0,"rho"]; theta1_mean = df1.loc[0,"theta mean"]
theta2 = df2.loc[0, "theta"]; rho2 = df2.loc[0,"rho"]; theta2_mean = df2.loc[0,"theta mean"]
theta3 = df3.loc[0, "theta"]; rho3 = df3.loc[0,"rho"]; theta3_mean = df3.loc[0,"theta mean"]
theta4 = df4.loc[0, "theta"]; rho4 = df4.loc[0,"rho"]; theta4_mean = df4.loc[0,"theta mean"]
```

```python
plt.figure(figsize=[10,7])
plt.plot(theta4,rho4*100,label='1 DMSO molecule (pre-equil), <theta>='+str(theta4_mean))
plt.plot(theta1,rho1*100,label='1 DMSO molecule (post-equil), <theta>='+str(theta1_mean))
plt.plot(theta2,rho2*100,label='9 DMSO molecules (post-equil), <theta>='+str(theta2_mean))
plt.plot(theta3,rho3*100,label='61 DMSO molecules (post-equil), <theta>='+str(theta3_mean))
plt.grid(True)
plt.xlabel('theta (S-O orientation), degrees')
plt.ylabel('Probability density x 100')
plt.legend()
```

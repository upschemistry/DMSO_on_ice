```python
# Parameters for the Gaussian
sigma = 10.0  # Standard deviation
size = 50     # Size of the kernel

# Create a Gaussian kernel
x = np.linspace(-size, size, 2 * size + 1)
gaussian_kernel = np.exp(-0.5 * (x / sigma) ** 2)
gaussian_kernel /= np.sum(gaussian_kernel)  # Normalize the kernel
plt.figure()
plt.plot(x,gaussian_kernel)
```

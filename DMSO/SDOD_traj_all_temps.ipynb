```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter1d
from matplotlib import colors
print(colors.CSS4_COLORS.keys())
```

```python
SDOD_224K = np.loadtxt("traj_224K_equilibrate/SDOD.txt")
SDOD_245K = np.loadtxt("traj_245K_equilibrate/SDOD.txt")
```

```python
print(np.shape(SDOD_224K))
time_224K = SDOD_224K[0,:]
SD_224K = SDOD_224K[1,:]
OD_224K = SDOD_224K[2,:]
time_245K= SDOD_245K[0,:]
SD_245K = SDOD_245K[1,:]
OD_245K = SDOD_245K[2,:]
smoothsize = 10000
SD_224K = uniform_filter1d(SD_224K,size=smoothsize)
OD_224K = uniform_filter1d(OD_224K,size=smoothsize)
SD_245K = uniform_filter1d(SD_245K,size=smoothsize)
OD_245K = uniform_filter1d(OD_245K,size=smoothsize)
```

```python
plt.figure()
plt.plot(time_224K,SD_224K, "darkcyan",label='SD 224K')
plt.plot(time_224K,OD_224K, "blue",label='OD 224K')
plt.plot(time_245K,SD_245K, "sienna",label='SD 245K')
plt.plot(time_245K,OD_245K, "red",label='OD 245K')
plt.xlabel('time (ns)')
plt.ylabel('distance (nm)')
plt.legend()
plt.grid(True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Use np.loadtxt to load in vac.xvg
data_last_frame = np.loadtxt('spectrum.xvg',comments=['@','#','&'])
data = np.loadtxt('../../DMSO/traj_245K_IRspectrum_after_equilibration/spectrum.xvg',comments=['@','#','&'])

# data_10 = np.loadtxt('spectrum_traj_whole_snapshots10.xvg',comments=['@','#','&'])
# data_9  = np.loadtxt('spectrum_traj_whole_snapshots9.xvg', comments=['@','#','&'])
# data_8  = np.loadtxt('spectrum_traj_whole_snapshots8.xvg', comments=['@','#','&'])
# data_7  = np.loadtxt('spectrum_traj_whole_snapshots7.xvg', comments=['@','#','&'])

# # Do some averaging
# stacked = np.array([\
#                     data_last_frame[:,1],
#                     data_10[:,1],
#                     data_9[:,1],
#                     data_8[:,1],
#                     data_7[:,1]
#                    ])
                  
# power_average = np.mean(stacked,axis=0)
```

```python
opacity = 0.8
linewidth = 2
nu_wavenum = data_last_frame[:,0]
nu_wavenum2 = data[:,0]
print('spectral resolution =',nu_wavenum[1]-nu_wavenum[0])
plt.figure
# plt.plot(nu_wavenum,power_average,color = 'k', linewidth=linewidth,label='average')
plt.plot(nu_wavenum,data_last_frame[:,1],color = 'blue', linewidth=linewidth, alpha=opacity, label='9 DMSO 245K')
plt.plot(nu_wavenum2,data[:,1],color = 'green', linewidth=linewidth, alpha=opacity, label='1 DMSO 245K')
# plt.plot(nu_wavenum,data_10[:,1],color = 'k', linewidth=linewidth, alpha=opacity, label='frame 10')
# plt.plot(nu_wavenum,data_9[:,1],color = 'k', linewidth=linewidth, alpha=opacity, label='frame 9')
# plt.plot(nu_wavenum,data_8[:,1],color = 'k', linewidth=linewidth, alpha=opacity, label='frame 8')
# plt.plot(nu_wavenum,data_7[:,1],color = 'k', linewidth=linewidth, alpha=opacity, label='frame 7')
plt.xlabel('wavenumber (cm-1)')
plt.ylabel('power (arb. units)')
plt.grid(True)
plt.xlim([500,1900])
# plt.ylim([0,200])
plt.legend()
```
